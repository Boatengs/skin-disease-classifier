# Skin Disease Classifier — Reviewed Results Walkthrough

This companion notebook documents the verified test results and modeling decisions from the EfficientNetB0 classifier. The project is a computer-vision research prototype and is **not a clinical diagnostic system**.


## 1. Make the dataset split explicit


```python
# The split matters because all headline performance should come from the held-out test set.
print("total_images: 10,015")
print("train: 8,012")
print("validation: 1,001")
print("test: 1,002")
print("classes: 7")
```

### What this tells us

The project separates training, validation, and test data instead of reporting performance on the data used to fit the model. The dataset is also strongly imbalanced, which is why the training pipeline uses class-weighted loss rather than assuming each class appears equally often.


## 2. Read the headline and class-specific results together


```python
# Overall accuracy should be interpreted alongside class-level behavior.
print("test_accuracy: 74.15%")
print("reported_melanoma_detection: 75.0%")
print("reported_vascular_lesion_detection: 92.9%")
```

### What this tells us

The overall test accuracy is **74.15%**, but performance is not uniform across lesion types. The repository reports stronger detection on vascular lesions than melanoma. Because the README labels these values as 'detection' rather than a specific statistic such as recall, this walkthrough keeps that terminology instead of silently relabeling the metric.


## 3. Connect architecture choices to the data problem


```python
# Transfer learning and weighting address limited data and class imbalance.
print("backbone: EfficientNetB0 pretrained on ImageNet")
print("trainable_parameters: 3.49M / 4.34M")
print("weighted_loss: CrossEntropyLoss")
print("augmentation: flip, rotate, color jitter")
print("explainability: Grad-CAM")
```

### What this tells us

Transfer learning supplies useful visual features without training a vision backbone from scratch. Weighted loss and augmentation directly address two common issues in medical imaging: uneven class frequency and limited examples. Grad-CAM adds transparency about **where the model looked**, but it does not prove that the model's reasoning is clinically valid.


## Final interpretation

The project demonstrates a complete classification workflow with held-out evaluation and visual explainability. The correct portfolio claim is that it is an explainable seven-class dermoscopy prototype—not that 74.15% accuracy is sufficient for clinical deployment.
